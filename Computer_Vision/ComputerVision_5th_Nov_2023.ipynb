# Let convert our color image to greyscale

```python
# Load an image
image = cv2.imread('Modi.jpg') # read the image
# View an image
cv2.imshow('Original Image', image) # write comment on the view image
cv2.waitKey()

# we use cvtColor, to convert to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2.imshow("Grayscale image", gray_image)
cv2.waitKey()

color_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
cv2.imshow("COLOR image", color_image)
cv2.waitKey()

cv2.destroyAllWindows()
```

```python
# Capture all color at one go
color_list = [method for method in dir(cv2) if method.startswith('COLOR_') is True]
color_list
len(color_list)

img = cv2.imread('opencv.jpg')

for color in range(len(color_list)):
    col_name = color_list[color]
    col_code = 'cv2.' + col_name
    resized = cv2.resize(img, (400,400))
    try:
        rgb_image = cv2.cvtColor(resized, eval(col_code))
        cv2.imshow(col_name + "_" + str(color), rgb_image)
        cv2.waitKey()
    except Exception as e:
        print("Error on :", str(col_code))
        pass
    cv2.destroyAllWindows()
```

# Drawing images and shapes using OpenCV/CV2

```python
import numpy as np
import cv2

# create a black image
image = np.zeros((512,512,3))
image1 = np.zeros((512,512))

cv2.imshow("Black rectangle (B&W)", image1)
cv2.waitKey(0)
cv2.imshow("Black rectangle (Color)", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Let's draw a line over black square

image1 = np.zeros((512,512,3))
cv2.line(image1, (0,0), (400,400), (0,255,255), 1)
cv2.imshow("Color Line", image1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Let's draw a rectangel over black square

image2 = np.zeros((512,512,3))
cv2.rectangle(image2, (100,100), (300,200), (255,0,255), 3)
cv2.imshow("Rectange", image2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Let's draw a square over black square
image3 = np.zeros((512,512,3))
cv2.rectangle(image3, (100,100), (200,200), (255,0,255), -1)
cv2.imshow("Square", image3)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Let's draw a Circle over black square
image4 = np.zeros((512,512,3))
cv2.circle(image4, (150,350), 100, (255,0,255), -1)
cv2.imshow("Circle", image4)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Let's draw a Circle over black square
image5 = np.zeros((512,512,3))
cv2.circle(image5, (150,350), 100, (255,0,255), -1)
cv2.imshow("Circle", image5)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
image6 = np.zeros((512,512,3))
cv2.putText(image6,"Hello", (75,350), cv2.FONT_HERSHEY_COMPLEX, 2, (0,0,255), 3)
cv2.imshow("Hello World!", image6)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
