```python
# Simple Maths behind image blending
# Image Blending = Weighted addition of matrices
img1.shape[0]
```

```python
img2_resized = cv.resize(img2, (img1.shape[1], img1.shape[0]))
```

```python
img2_resized.shape
```

```python
# Now you use addWeighted with img1 and img2_resized
# alpha = 0.3
# cv2.addWeighted(img1, alpha, img2, beta, gamma)
# here alpha means (1-alpha), beta=alpha, gamma - scaler value which will add on the pixel cell value
dst = cv.addWeighted(img1, 0.7, img2_resized,0.3,0)
cv2_imshow(dst)
```
