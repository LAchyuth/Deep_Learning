# Case Study: IMDB movie review sentiment classification dataset

```python
from keras.datasets import imdb  # inbuilt dataset
from keras import Sequential
from keras.layers import Dense, SimpleRNN, Embedding, Flatten
```

```python
(x_train,y_train),(x_test,y_test) = imdb.load_data()
```

```python
print(x_train.shape,y_train.shape),(x_test.shape,y_test.shape)
```

```python
x_train[2]
```

```python
print(len(x_train[0]))
print(len(x_train[1]))
print(len(x_train[5]))
```

```python
# padding required as we have different lengths or timesteps in each review
x_train =pad_sequences(x_train, padding ='post',maxlen =200) # we can use max_length to control no of words required
x_test = pad_sequences(x_test, padding ='post',maxlen =200)
```

```python
print(len(x_train[0]))
print(len(x_train[1]))
print(len(x_train[5]))
```

# Simple RNN Model-Interger Encoding

```python
model = Sequential()
model.add(SimpleRNN(32, input_shape =(200,1),return_sequences =False))
#return_sequences =False = many to one type rnn
model.add(Dense(1,activation ='sigmoid'))
model.summary()
```

```python
print((1*32+(32*32)+32))
print((32*1+1))
```

```python
model.compile(optimizer ='adam',loss ='binary_crossentropy',metrics =['accuracy'])
model.fit(x_train,y_train,epochs =10,batch_size =32,validation_data =(x_test,y_test))
```

```python
# here i got pathetic accuracy because of 2 reasons
 #1.out of 2494 time steps, i considered max words as 200, so we don't get proper meaning by reducing the length
 #2. I used integer encoding as against embedding encoding
```

# Simple RNN Model-Interger Encoding-with max vocab

```python
# padding required as we have different lengths or timesteps in each review
x_train1 =pad_sequences(x_train, padding ='post',maxlen =2494)
x_test1 = pad_sequences(x_test, padding ='post',maxlen =2494)
```

```python
model = Sequential()
model.add(SimpleRNN(32, input_shape =(2494,1),return_sequences =False))
#return_sequences =False = many to one type rnn
model.add(Dense(1,activation ='sigmoid'))
model.summary()
```

```python
model.compile(optimizer ='adam',loss ='binary_crossentropy',metrics =['accuracy'])
model.fit(x_train1,y_train,epochs =5,batch_size =32,validation_data =(x_test1,y_test))
```

#Approach 2-Embedding Method(Dense Representation)

```python
model = Sequential()
model.add(Embedding(10000,2))
model.add(SimpleRNN(32,return_sequences =False))
model.add(Dense(1,activation ='sigmoid'))
model.summary()
```

```python
model.compile(optimizer ='adam',loss ='binary_crossentropy',metrics =['accuracy'])
model.fit(x_train,y_train,epochs =5,batch_size =32,validation_data =(x_test,y_test))
```
