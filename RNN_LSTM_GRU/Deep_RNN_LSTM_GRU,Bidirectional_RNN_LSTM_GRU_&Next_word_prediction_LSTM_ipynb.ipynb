```python

import tensorflow as tf
from tensorflow import keras
from keras.preprocessing.text import Tokenizer
```

```python
tokenizer =Tokenizer()
tokenizer.fit_on_texts([my_paragraph])
```

```python
len(tokenizer.word_index)
# unique vocab size
```

```python
input_sequence =[]
for sentence in my_paragraph.split('\n'):
  tokenized_sentence =tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1,len(tokenized_sentence)):
    input_sequence.append(tokenized_sentence[:i+1])
```

```python
input_sequence
```

```python
max_len =max([len(x) for x in input_sequence])
```

```python
max_len
#max words in a sentence
```

```python
from keras.preprocessing.sequence import pad_sequences
pad_input_sequence = pad_sequences(input_sequence,maxlen =max_len,padding ='pre')
# each sentence has different length, making common for all using pre padding
```

```python
pad_input_sequence
```

```python
x =pad_input_sequence[:,:-1] # because we are using last word prediction, so last word is target word
y =pad_input_sequence[:,-1]
```

```python
x.shape
```

```python
y.shape
```

```python
from keras.utils import to_categorical
y = to_categorical(y,num_classes=200) # randomly we kept categorical classes is 200
                                      # that many outputs words
```

# Building LSTM Model


```python
from keras import Sequential
from keras.layers import Embedding,LSTM,Dense
from keras.models import Model
from keras import Input
```

```python
# Define model(another method of writing lstm code)

timesteps =40
features =3
LSTMoutputdimension =2

input =Input(shape=(timesteps,features))
output = LSTM(LSTMoutputdimension)(input)
model_LSTM = Model(inputs = input, outputs =output)
model_LSTM.summary()

 # 4[(2+3)*2+2] =48  # here 4 parameters from lstm
 # for reference
```

```python
model = Sequential()
model.add(LSTM(150, input_shape =(timesteps, features)))

# for reference
```

```python
model =Sequential()
model.add(Embedding(200,100, input_length =25)) # x shape & here activation will be taken automatically
model.add(LSTM(150))                        # 100 is neurons & 200 is first 200 words from y
# whenever we write embedding, we have to write input shape(100 in this case)
model.add(Dense(200, activation ='softmax'))
model.compile(loss = 'categorical_crossentropy',optimizer ='adam',metrics =['accuracy'])
model.summary()

#4[(150+100)*150+150]
```

```python
"""
Embedding Layer:

Input dimension: 250 (vocabulary size)
Output dimension: 100
Parameters: 250 (input dimension) * 100 (output dimension) = 25,000
LSTM Layer:

Number of LSTM units: 150
Parameters:
Weights: 4 * 100 (input dimension) * 150 (number of units) = 60,000
Recurrent weights: 4 * 150 (number of units) * 150 (number of units) = 90,000
Biases: 4 * 150 (number of units) = 600
Total parameters for LSTM layer: 60,000 + 90,000 + 600 = 150,600
Dense Layer:

Input size (coming from LSTM layer): 150
Output size: 200 (number of classes)
Parameters: 150 (input size) * 200 (output size) = 30,000
Total Parameters:

Embedding Layer: 25,000
LSTM Layer: 150,600
Dense Layer: 30,000
Total: 25,000 + 150,600 + 30,000 = 205,800
"""
```

```python
model.fit(x,y,epochs=100)
```

```python
# predict the next word

import numpy as np
import time

text ="macine learning"  # first word of sentence

for i in range(15):
  token_text = tokenizer.texts_to_sequences([text])[0]
  #padding
  padded_token_text = pad_sequences([token_text],maxlen =25,padding ='pre')
  #predict
  predict_sentence =np.argmax(model.predict(padded_token_text))

  for word, index in tokenizer.word_index.items():
    if index == predict_sentence:
      text = text + " " + word
      print(text)
      time.sleep(2)

# from the above sentence in the 1st cell, if we give any word automatically it will create a story till the given range
```

#Deep _RNN / DeeP _LSTM / Deep_GRU

```python
import tensorflow as tf
from tensorflow import keras
from keras.datasets import imdb
from keras.preprocessing.sequence import pad_sequences
from keras.models import Sequential
from keras.layers import Embedding, SimpleRNN, Dense, LSTM, GRU, Bidirectional
```

```python
# Load the dataset
(x_train,y_train),(x_test,y_test) = imdb.load_data(num_words =10000)
x_train =pad_sequences(x_train, maxlen =100)
x_test = pad_sequences(x_test, maxlen =100)
```

# Deep RNN

```python
model1 = Sequential([
    Embedding(10000,32, input_length =100),
    SimpleRNN(5, return_sequences = True),
    SimpleRNN(5),
    Dense(1,activation ='sigmoid') #Dense =1 means we get only 1output either 1 or 0
])

model1.summary()
```

# Deep LSTM

```python
model2 = Sequential([
    Embedding(10000,32, input_length =100),
    LSTM(5, return_sequences = True),
    LSTM(5),
    Dense(1,activation ='sigmoid')
])

model2.summary()
```

# Deep GRU

```python
model3 = Sequential([
    Embedding(10000,32, input_length =100),
    GRU(5, return_sequences = True),
    GRU(5),
    Dense(1,activation ='sigmoid')
])

model3.summary()
```

```python
model3.compile(optimizer ='adam', loss ='binary_crossentropy',metrics =['accuracy'])
```

```python
history = model3.fit(x_train,y_train,epochs =5, batch_size =32,validation_split =0.2)
```

# Bi-Directional LSTM

```python
model4 = Sequential([
    Embedding(10000,32,input_length =100),
    Bidirectional(LSTM(5,return_sequences =True)),
    Dense(1,activation = 'sigmoid')
])
model4.summary()
```

```python
model4.compile(optimizer ='adam', loss ='binary_crossentropy',metrics =['accuracy'])
```
